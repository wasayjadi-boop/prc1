Q1. Write a program that takes two numbers (a = 15 , b = 4) and prints their:


- Sum

- Difference

- Product

- Division

- Floor Division

- Modulus

```python
a=15
b=4
sum=a+b
diff=a-b
pro=a*b
divi=a/b
floordivision = a//b
modulus=a%b

print("Sum=",sum)
print("Difference=",diff)
print("Product=",pro)
print("Division=",divi)
print("Floor_division=",floordivision,)
print("Modulus=",modulus)
```
